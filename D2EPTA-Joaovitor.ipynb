```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- 1. CARREGAMENTO E LIMPEZA DE DADOS ---
# Carrega o arquivo original (sem forçar decimal, para testar a limpeza manual)
try:
    df = pd.read_csv('nashira_telemetry.csv') # Lê como string se tiver aspas/vírgulas
except FileNotFoundError:
    print("❌ Erro: Arquivo 'nashira_telemetry.csv' não encontrado.")
    # Pare a execução se não achar o arquivo

# Lista de colunas numéricas que precisam de limpeza
cols_to_convert = [
    'lat', 'lon', 'alt_gps', 'alt_bar', 'pressure', 'temperature',
    'pos_n', 'pos_e', 'pos_d', 'vel_n', 'vel_e', 'vel_d',
    'acc_n', 'acc_e', 'acc_d', 'acc_x_raw', 'acc_y_raw', 'acc_z_raw',
    'gyro_x', 'gyro_y', 'gyro_z', 'mag_x', 'mag_y', 'mag_z',
    'roll', 'pitch', 'yaw'
]

def clean_and_convert(val):
    """Remove aspas e substitui vírgula por ponto para converter string em float."""
    if isinstance(val, str):
        val = val.replace('"', '').replace(',', '.')
        try:
            return float(val)
        except ValueError:
            return np.nan
    return val

# Aplica a conversão em todas as colunas listadas (se existirem)
for col in cols_to_convert:
    if col in df.columns:
        df[col] = df[col].apply(clean_and_convert)

# Garante que o timestamp seja numérico
df['timestamp_ms'] = pd.to_numeric(df['timestamp_ms'], errors='coerce')

# Criação de uma coluna de tempo relativo em segundos (começando do zero)
df['time_sec'] = (df['timestamp_ms'] - df['timestamp_ms'].iloc[0]) / 1000.0

# --- 2. CÁLCULOS DE ENGENHARIA ---

# Cálculo da Velocidade e Aceleração Resultante (Magnitude vetorial)
df['vel_total'] = np.sqrt(df['vel_n']**2 + df['vel_e']**2 + df['vel_d']**2)
df['acc_total'] = np.sqrt(df['acc_x_raw']**2 + df['acc_y_raw']**2 + df['acc_z_raw']**2)

# Cálculo Alternativo de Velocidade (Derivada Barométrica)
df['vel_baro'] = df['alt_bar'].diff() / df['time_sec'].diff()
# Suavização (Média Móvel)
df['vel_baro_smooth'] = df['vel_baro'].rolling(window=10).mean()

# --- 3. DETECÇÃO DE EVENTOS DE VOO ---

# Detectar Lançamento
df['acc_smooth'] = df['acc_total'].rolling(window=5).mean()
# Procura aceleração > 1.5G
liftoff_candidates = df[df['acc_smooth'] > 1.5]

if not liftoff_candidates.empty:
    liftoff_idx = liftoff_candidates.index.min()
else:
    # Fallback: Se não achar aceleração alta, usa mudança de altitude > 5m
    altitude_candidates = df[df['alt_bar'] > (df['alt_bar'].iloc[0] + 5)]
    if not altitude_candidates.empty:
        liftoff_idx = altitude_candidates.index.min()
    else:
        liftoff_idx = df.index[0] # Se nada acontecer, assume o começo

liftoff_time = df.loc[liftoff_idx, 'time_sec']

# Detectar Apogeu
apogee_idx = df['alt_bar'].idxmax()
apogee_time = df.loc[apogee_idx, 'time_sec']
max_alt = df['alt_bar'].max()
start_alt = df['alt_bar'].iloc[0]

# Definir janela de plotagem
start_plot = max(0, liftoff_time - 5)
end_plot = df['time_sec'].max()

# Filtrar o DataFrame apenas para a janela de voo
flight_df = df[(df['time_sec'] >= start_plot) & (df['time_sec'] <= end_plot)].copy()

# --- 4. GERAÇÃO DOS GRÁFICOS ---
if not flight_df.empty:
    fig, axes = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

    # Gráfico 1: Altitude
    axes[0].plot(flight_df['time_sec'], flight_df['alt_bar'], color='#1f77b4', linewidth=2)
    axes[0].axvline(liftoff_time, color='green', linestyle='--', label='Lançamento')
    axes[0].axvline(apogee_time, color='orange', linestyle='--', label='Apogeu')
    axes[0].set_ylabel('Altitude (m)')
    axes[0].set_title(f'Perfil de Altitude (Apogeu: {max_alt - start_alt:.1f}m AGL)')
    axes[0].legend()
    axes[0].grid(True, linestyle='--', alpha=0.7)

    # Gráfico 2: Velocidade Vertical
    axes[1].plot(flight_df['time_sec'], flight_df['vel_baro_smooth'], color='#ff7f0e', linewidth=2, label='Vel. Barométrica')
    axes[1].set_ylabel('Velocidade (m/s)')
    axes[1].set_title('Velocidade Vertical Estimada')
    axes[1].axhline(0, color='black', linewidth=1)
    axes[1].legend()
    axes[1].grid(True, linestyle='--', alpha=0.7)

    # Gráfico 3: Aceleração
    axes[2].plot(flight_df['time_sec'], flight_df['acc_total'], color='#d62728', linewidth=1.5, label='Aceleração Total')
    axes[2].set_ylabel('Aceleração (G)')
    axes[2].set_xlabel('Tempo (s)')
    axes[2].set_title('Aceleração Resultante')
    axes[2].grid(True, linestyle='--', alpha=0.7)

    plt.tight_layout()
    plt.savefig('analise_voo_nashira_final.png')
    plt.show()

    # Imprimir métricas no console
    print(f"--- Relatório de Voo ---")
    print(f"Tempo de Lançamento: {liftoff_time:.2f} s")
    print(f"Tempo de Apogeu: {apogee_time:.2f} s")
    print(f"Duração da Subida: {apogee_time - liftoff_time:.2f} s")
    print(f"Altitude Máxima (AGL): {max_alt - start_alt:.2f} m")
    try:
        print(f"Velocidade Máxima (Est.): {flight_df['vel_baro_smooth'].max():.2f} m/s")
        print(f"Aceleração Máxima: {flight_df['acc_total'].max():.2f} G")
    except:
        pass
else:
    print("⚠️ Aviso: Nenhum dado de voo encontrado na janela selecionada.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Garante que a coluna de tempo existe com o nome certo 'tempo_s'
if 'timestamp_ms' in df.columns:
    df['tempo_s'] = (df['timestamp_ms'] - df['timestamp_ms'].iloc[0]) / 1000.0
elif 'time_sec' in df.columns:
    df['tempo_s'] = df['time_sec'] # Usa a que já existe

# Garante que a altitude seja número (troca vírgula por ponto se necessário)
# Isso evita erros se você carregou o arquivo e pulou a limpeza
df['alt_bar'] = df['alt_bar'].astype(str).str.replace(',', '.')
df['alt_bar'] = pd.to_numeric(df['alt_bar'], errors='coerce')

# --- 1. CÓDIGO DE ANÁLISE ---

# Limpeza de Ruído (Suavização)
df['alt_suave'] = df['alt_bar'].rolling(window=20, center=True).mean()

# Detecção Automática de Eventos
# Encontra o índice (linha) onde a altitude foi máxima
idx_apogeu = df['alt_suave'].idxmax()
tempo_apogeu = df['tempo_s'].iloc[idx_apogeu]
alt_apogeu = df['alt_suave'].iloc[idx_apogeu]

# --- 2. PLOTAGEM ---
plt.figure(figsize=(12, 6))

# Plota os dados brutos (cinza claro)
plt.plot(df['tempo_s'], df['alt_bar'], color='gray', label='Dados Brutos (Sensor)', alpha=0.5)

# Plota os dados limpos (azul forte)
plt.plot(df['tempo_s'], df['alt_suave'], color='blue', linewidth=2, label='Dados Tratados (Suavizados)')

# Adiciona anotação no Apogeu
plt.annotate(f'Apogeu: {alt_apogeu:.1f}m \n t={tempo_apogeu:.1f}s',
             xy=(tempo_apogeu, alt_apogeu),
             xytext=(tempo_apogeu + 10, alt_apogeu + 10),
             arrowprops=dict(facecolor='red', shrink=0.05))

plt.title('Análise Detalhada do Voo: Identificação de Eventos e Tratamento de Dados')
plt.xlabel('Tempo (s)')
plt.ylabel('Altitude (m)')
plt.legend()
plt.grid(True)
plt.show()

# Imprimir conclusão
print(f"ANÁLISE DO VOO:")
print(f"O apogeu detectado foi de {alt_apogeu:.2f} metros.")
print(f"Isso ocorreu aos {tempo_apogeu:.2f} segundos de voo.")
```
